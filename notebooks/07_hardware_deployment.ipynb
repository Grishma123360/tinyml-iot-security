```python
import os

# 1. Establish project directory paths
base_dir = r"C:\Users\ACER\tinyml-iot-security"
tflite_path = os.path.join(base_dir, "models", "freeze_replay_detector_quant.tflite")
header_output_path = os.path.join(base_dir, "models", "freeze_replay_detector.h")

print(f"Reading quantized TFLite binary from: {tflite_path}")

# 2. Open binary file and stream data directly into a buffered text writer
file_size = os.path.getsize(tflite_path)

with open(tflite_path, 'rb') as bin_file, open(header_output_path, 'w', buffering=2048) as out_file:
    # Write C++ macro guards and headers
    out_file.write("// TinyML Freeze/Replay Attack Detector Model\n")
    out_file.write("// Generated automatically for high-speed embedded hardware deployment\n\n")
    out_file.write("#ifndef FREEZE_REPLAY_DETECTOR_H\n")
    out_file.write("#define FREEZE_REPLAY_DETECTOR_H\n\n")
    out_file.write("const unsigned char g_freeze_replay_detector_model[] __attribute__((aligned(8))) = {\n  ")
    
    # Process file chunks using optimized byte slicing
    byte_counter = 0
    while (byte := bin_file.read(1)):
        # Convert raw byte directly to standard hex signature
        out_file.write(f"0x{byte[0]:02x}")
        byte_counter += 1
        
        if byte_counter < file_size:
            # Format row elements with clean spacing anchors
            if byte_counter % 12 == 0:
                out_file.write(",\n  ")
            else:
                out_file.write(", ")

    # Close out arrays and log static length metrics
    out_file.write("\n};\n\n")
    out_file.write(f"const unsigned int g_freeze_replay_detector_model_len = {file_size};\n\n")
    out_file.write("#endif // FREEZE_REPLAY_DETECTOR_H\n")

print(f"\n[Success] Instant compilation complete! Saved to: {header_output_path}")
print(f"Total Model Memory Footprint on Flash: {file_size} bytes (~{file_size/1024:.2f} KB)")
```
